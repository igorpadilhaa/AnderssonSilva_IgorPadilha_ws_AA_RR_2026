```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_json('resultados_benchmark.jsonl', lines=True)
df['algoritmo'] = 'Recursivo'

df_avg = df.groupby(['algoritmo', 'n'])['tempo_ns'].mean().reset_index()
df_pivot = df_avg.pivot(index='n', columns='algoritmo', values='tempo_ns')

df_pivot.plot(kind='line', marker='o', figsize=(10, 6))

plt.title('Benchmark Fibonacci: Comparação de Algoritmos')
plt.xlabel('n (Posição na sequência)')
plt.ylabel('Tempo Médio (nanosegundos)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Algoritmo')
plt.xticks(sorted(df['n'].unique())) 

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_json('resultados_benchmark.jsonl', lines=True)
df['algoritmo'] = 'Recusivo'

df_avg = df.groupby(['algoritmo', 'n'])['tempo_ns'].mean().reset_index()
df_pivot = df_avg.pivot(index='n', columns='algoritmo', values='tempo_ns')

plt.figure(figsize=(10, 6))

for algo in df_pivot.columns:
    serie = df_pivot[algo].dropna()
    plt.plot(serie.index, serie.values, marker='o', label=algo)

plt.xscale('log')
plt.yscale('log')

# 4. Ajustes estéticos
plt.title('Benchmark de Fibonacci Recusivo')
plt.xlabel('n (Escala Logarítmica)')
plt.ylabel('Tempo Médio (ns - Escala Logarítmica)')
plt.grid(True, which="both", linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df_avg = df.groupby(['algoritmo', 'n'])['tempo_ns'].mean().reset_index()
df_pivot = df_avg.pivot(index='n', columns='algoritmo', values='tempo_ns')

plt.figure(figsize=(12, 6))

for algo in df_pivot.columns:
    serie = df_pivot[algo].dropna()
    plt.plot(serie.index, serie.values, marker='o', label=algo)

plt.xscale('log')
plt.yscale('log')

valores_n = sorted(df['n'].unique())

plt.xticks(valores_n, labels=[str(v) for v in valores_n], rotation=45)

plt.title('Benchmark Fibonacci: Performance por Termo (n)')
plt.xlabel('n (Valores de teste)')
plt.ylabel('Tempo Médio (ns - Escala Log)')
plt.grid(True, which="major", linestyle='-', alpha=0.5)
plt.grid(True, which="minor", linestyle=':', alpha=0.2)
plt.legend()
plt.tight_layout()

plt.show()
```
